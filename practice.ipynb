### SINGLE LINKED LIST

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
```

```python
node1=Node(10)
node2=Node(20)
node3=Node(30)

node1.next=node2
node2.next=node3
```

```python
print(node1.data)
```

```python
print(node1.next.data)
```

```python
#DISPLAY
current=node1
while current is not None:
    print(current.data, end=" -> ")
    current=current.next

print("None")
```

```python
#INSERT AT BEGINING
new_node=Node(5) 

new_node.next=node1   #connecxting new node to the head position

node1=new_node       #Make the new node the head
```

```python
current = node1

while current is not None:
    print(current.data, end=" → ")
    current = current.next

print("None")
```

```python
def insert_beginning(head,data):        #making it as a function

    new_node=Node(data)

    new_node.next=head

    head=new_node

    return head
```

```python
node1=insert_beginning(node1,3)
```

```python
current = node1

while current is not None:
    print(current.data, end=" → ")
    current = current.next

print("None")
```

```python
#INSERT AT END
new_node=Node(40)

current = node1

while current.next is not None:
    current = current.next

current.next=new_node
```

```python
current = node1

while current is not None:
    print(current.data, end=" → ")
    current = current.next

print("None")
```

```python
def insert_end(head, data):

    new_node = Node(data)

    if head is None:
        return new_node

    current = head

    while current.next is not None:
        current = current.next

    current.next = new_node

    return head
```

```python
node1=insert_end(node1,50)
```

```python
current=node1

while current.next is not None:
    print(current.data, end=" -> ")
    current=current.next

print("None")
```

```python
#insertion at a specific position

def insert_after(head, target, data):

    current = head

    while current is not None:

        if current.data == target:

            new_node = Node(data)

            new_node.next = current.next

            current.next = new_node

            return head

        current = current.next

    print("Target not found")

    return head
```

```python
node1=insert_after(node1,10,15)
```

```python
current=node1

while current.next is not None:
    print(current.data, end=" -> ")
    current=current.next

print("None")
```

```python
def search(head,value):
    current=head
    while current.next is not None:
        if current.data==value:
            return True

        current=current.next

    return False
```

```python
print(search(node1,15))
```

```python
print(search(node1, 100))
```
