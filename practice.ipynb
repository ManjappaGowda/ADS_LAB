### SINGLE LINKED LIST

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
```

```python
node1=Node(10)
node2=Node(20)
node3=Node(30)

node1.next=node2
node2.next=node3
```

```python
print(node1.data)
```

```python
print(node1.next.data)
```

```python
#DISPLAY
current=node1
while current is not None:
    print(current.data, end=" -> ")
    current=current.next

print("None")
```

```python
#INSERT AT BEGINING
new_node=Node(5) 

new_node.next=node1   #connecxting new node to the head position

node1=new_node       #Make the new node the head
```

```python
current = node1

while current is not None:
    print(current.data, end=" → ")
    current = current.next

print("None")
```
